```python
# Calculations:

def mean(step):
    up = (step[0]+step[1])/2
    low = (step[2]+step[3])/2
    return up, low

def variance(step):
    up = np.abs(step[0]-step[1])/2
    low = np.abs(step[2]-step[3])/2
    return up , low

def distance(step):
    d = np.abs(mean(step)[0]-mean(step)[1])
    err = variance(step)[0]+variance(step)[1]
    return (d, err)


def result(steps):
    results = []
    for i in range(48):
        results.append(np.round(distance(steps[i]),2))
    return np.array(results)


m = result(steps_arr)
order = np.argsort(m[:,0])
m_sorted = m[order]
print(m_sorted)

```
